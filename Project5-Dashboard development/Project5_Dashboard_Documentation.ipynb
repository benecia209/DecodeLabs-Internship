# Project 5: Retail Sales Dashboard Development

## 1. Project Overview

This project focuses on developing an interactive retail sales dashboard using Microsoft Power BI. The purpose of the project is to transform retail sales data into meaningful visual insights that can help management understand business performance.

The dashboard presents key performance indicators (KPIs), sales trends, profit performance, customer segments, product performance, regional performance, and order-related information.

## 2. Problem Statement

Retail businesses generate large amounts of sales data. Analyzing this data manually can make it difficult to identify sales trends, understand product performance, monitor profitability, and compare different business segments.

The objective of this project is to develop an interactive dashboard that organizes sales data into clear visualizations and enables users to explore business performance through filters.

## 3. Project Objectives

* To develop an interactive retail sales dashboard using Power BI.
* To calculate and display important business KPIs.
* To analyze sales and profit across different categories and regions.
* To identify sales trends over time.
* To visualize customer segment and product performance.
* To provide interactive filters for exploring business data.
* To present data in a clear and understandable format to support business decisions.

A synthetic retail sales dataset was prepared specifically for this project.

## 6. Methodology

### Step 1: Data Import

The retail sales dataset was imported into Power BI using the Excel workbook connector. The Sales_Data worksheet was selected and loaded into the report.

### Step 2: Data Type Verification

The data types of the columns were checked to ensure that dates, numerical values, and categorical fields were interpreted correctly.

### Step 3: KPI Creation

Four DAX measures were created to calculate the main business performance indicators.

**Total Sales**

```DAX
Total Sales = SUM(Sales_Data[Sales])
```

**Total Profit**

```DAX
Total Profit = SUM(Sales_Data[Profit])
```

**Total Orders**

```DAX
Total Orders = DISTINCTCOUNT(Sales_Data[OrderID])
```

**Total Quantity**

```DAX
Total Quantity = SUM(Sales_Data[Quantity])
```

### Step 4: Dashboard Development

Different Power BI visualizations were created to analyze various aspects of the retail business.

The dashboard includes:

* KPI cards for sales, profit, total orders, and quantity.
* A line chart for monthly sales trends.
* Bar charts for sales and profit by category and region.
* A product sales chart.
* A customer segment chart.
* A doughnut chart for order status distribution.
* Charts for payment and shipping methods.
* A scatter chart for comparing sales and profit by category.

### Step 5: Interactive Filters

Slicers were added to allow users to filter the dashboard by:

* Region
* Category
* Customer Segment
* Order Status
* Order Date

These filters allow users to explore selected portions of the dataset.

### Step 6: Dashboard Formatting

The dashboard layout, chart titles, KPI number formats, and visual sizes were adjusted to improve readability and consistency.

* **Total Profit:** Approximately 63.4K

* **Total Quantity:** Approximately 2,900

The dashboard also provides the following visual observations:

* Technology contributes the largest share of sales among the displayed product categories.
* The Consumer segment contributes the largest share of sales in the displayed customer segment chart.
* Sales and profit vary across regions and product categories.
* The monthly sales chart shows noticeable changes in sales throughout the year.
* The order status chart displays the distribution of orders among the available statuses.

These observations are based on the dashboard's displayed visuals and should be interpreted in the context of the synthetic dataset.

## 9. Business Recommendations

Based on the dashboard's visual analysis, the following actions could be considered:

* Monitor sales trends to identify months with higher and lower sales.
* Review the performance of product categories to understand their contribution to overall sales.
* Investigate regions with comparatively lower sales or profit.
* Analyze customer segments to understand differences in purchasing contributions.
* Review order status and shipping method distributions to identify areas that may need operational attention.
* Use the dashboard's filters to investigate specific products, regions, customer segments, and time periods before making business decisions.

These recommendations are exploratory and should be validated with additional operational and financial information before implementation.
