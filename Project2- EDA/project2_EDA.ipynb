# Day 4 – Outlier Analysis

## Objective

Outlier analysis was performed to identify unusually high or low values
in the numerical variables.

## Method

The Interquartile Range (IQR) method was used to identify potential
outliers.

The outlier boundaries were calculated using:

Lower Bound = Q1 - 1.5 × IQR

Upper Bound = Q3 + 1.5 × IQR

## TotalPrice

Potential outliers were identified in the TotalPrice column using the IQR
method. These values represent orders that are unusually high or low
compared with the majority of orders.

The identified records were inspected using OrderID, Product, Quantity,
UnitPrice, ItemsInCart, and TotalPrice to determine whether they appeared
to be genuine transactions.

## UnitPrice

Potential outliers were also identified in the UnitPrice column. These
values were investigated rather than automatically removed because a high
unit price may represent a legitimate expensive product.

## Quantity and ItemsInCart

Box plots were used to visually examine Quantity and ItemsInCart for
unusual observations.

## Conclusion

Outlier analysis helps identify unusual observations in the dataset.
However, an outlier is not necessarily an error. Therefore, potential
outliers should be investigated before deciding whether they should be
removed or retained.

```python
df[["Quantity", "UnitPrice", "ItemsInCart", "TotalPrice"]].hist(
    figsize=(12, 8),
    bins=20
)

plt.tight_layout()
plt.show()
```

```python
Q1 = df["TotalPrice"].quantile(0.25)
Q3 = df["TotalPrice"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print(Q1, Q3, IQR, lower_bound, upper_bound)
```

```python
outliers_totalprice = df[
    (df["TotalPrice"] < lower_bound) |
    (df["TotalPrice"] > upper_bound)
]

print(len(outliers_totalprice))
```

```python
outliers_totalprice[
    ["OrderID", "Product", "Quantity", "UnitPrice", "ItemsInCart", "TotalPrice"]
].sort_values("TotalPrice", ascending=False)
```

```python
numerical_columns = ["Quantity", "UnitPrice", "ItemsInCart", "TotalPrice"]

for column in numerical_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    
    outliers = df[(df[column] < lower) | (df[column] > upper)]
    
    print(column)
    print("Lower bound:", lower)
    print("Upper bound:", upper)
    print("Number of outliers:", len(outliers))
    print("-" * 40)
```
